```python
n = int(input("enter number of elements: "))

numbers = []

for i in range(n):
    value = int(input("enter number: "))
    numbers.append(value)

print("numbers in list:")
for num in numbers:
    print(num)

even_count = 0
odd_count = 0

for num in numbers:
    if num % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("even numbers count:", even_count)
print("odd numbers count:", odd_count)

index = int(input("enter index to replace: "))
new_value = int(input("enter new value: "))

numbers[index] = new_value

print("updated list:", numbers)
```

```python
def totalsalary(salary):
    hra = salary * 10 / 100
    da = salary * 5 / 100
    total = salary + hra + da
    return total

basic = int(input("enter basic salary: "))
result = totalsalary(basic)

print("total salary:", result)
```

```python
def stringinfo(text):
    total = len(text)
    upper = 0
    lower = 0
    digit = 0

    for ch in text:
        if ch.isupper():
            upper = upper + 1
        elif ch.islower():
            lower = lower + 1
        elif ch.isdigit():
            digit = digit + 1

    result = {
        "total characters": total,
        "uppercase letters": upper,
        "lowercase letters": lower,
        "digits": digit
    }
    return result


usertext = input("enter a string: ")
info = stringinfo(usertext)
print(info)
```

```python
text = input("enter a string: ")

maxchar = ""
maxcount = 0

currentchar = ""
currentcount = 0

for ch in text:
    if ch == currentchar:
        currentcount = currentcount + 1
    else:
        currentchar = ch
        currentcount = 1

    if currentcount > maxcount:
        maxcount = currentcount
        maxchar = currentchar

longestsubstring = maxchar * maxcount

print("longest substring:", longestsubstring)
print("length:", maxcount)
```

```python
def findrepeated(numbers):
    repeated = []
    checked = []

    for num in numbers:
        if num not in checked:
            count = 0
            for x in numbers:
                if x == num:
                    count = count + 1

            if count > 1:
                repeated.append(num)

            checked.append(num)

    return repeated


userlist = input("enter numbers separated by space: ").split()
numbers = []
for item in userlist:
    numbers.append(int(item))

result = findrepeated(numbers)
print(result)
```
