```python
class SecurityAgent:
    def __init__(self, agent_id, name, status="inactive"):
        self.agent_id = agent_id
        self.name = name
        self.__status = status 

    def get_status(self):
        return self.__status

    def set_status(self, status):
        self.__status = status

class FirewallAgent(SecurityAgent):
    def monitor_traffic(self):
        self.set_status("active")
        print(f"{self.name} is monitoring network traffic.")

class MalwareDetectionAgent(SecurityAgent):
    def scan_files(self):
        self.set_status("Active")
        print(f"{self.name} is scanning files for malware.")

class AutomationAgent(SecurityAgent):
    def run_automation(self):
        self.set_status("Active")
        print(f"{self.name} is performing ai automation tasks")

firewall = FirewallAgent(101, "firewall Mohammad Burair")
malware = MalwareDetectionAgent(102, "malware Zahran Ali")
automation = AutomationAgent(103, "autobot maaz")

firewall.monitor_traffic()
malware.scan_files()
automation.run_automation()
```

```python
class Threat:
    def __init__(self, threat_id, name, severity):
        self.threat_id = threat_id
        self.name = name
        self.__severity = severity

    def get_severity(self):
        return self.__severity

    def set_severity(self, severity):
        self.__severity = severity

class PhishingThreat(Threat):
    def analyze_email(self):
        self.set_severity("high")
        print(f"{self.name} is analyzing suspicious emails")

class RansomwareThreat(Threat):
    def scan_files(self):
        self.set_severity("critical")
        print(f"{self.name} is scanning files for ransomware")

class BotnetThreat(Threat):
    def detect_traffic(self):
        self.set_severity("medium")
        print(f"{self.name} is detecting unusual network traffic")

phishing = PhishingThreat(201, "phishing alert", "medium")
ransomware = RansomwareThreat(202, "ransomware alert", "high")
botnet = BotnetThreat(203, "botnet alert", "low")

phishing.analyze_email()
ransomware.scan_files()
botnet.detect_traffic()
```

```python
class ResponseAgent:
    def execute_response(self):
        print("executing generic response")

class AlertAgent(ResponseAgent):
    def execute_response(self):
        print("sending alert notifications")

class BlockAgent(ResponseAgent):
    def execute_response(self):
        print("blocking malicious activities")

class RecoverAgent(ResponseAgent):
    def execute_response(self):
        print("restoring affected systems")

agents = [AlertAgent(), BlockAgent(), RecoverAgent()]

for agent in agents:
    agent.execute_response()
```

```python
class Employee:
    def work(self):
        print("employee is working")

class Manager(Employee):
    def work(self):
        print("manager is planning and managing tasks")

class Developer(Employee):
    def work(self):
        print("developer is writing code")

class Designer(Employee):
    def work(self):
        print("designer is creating designs")

employees = [Manager(), Developer(), Designer()]

for employee in employees:
    employee.work()
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.__grade = None

    def set_grade(self, grade):
        self.__grade = grade

    def get_grade(self):
        return self.__grade

    def display_info(self):
        print(f"{self.name} has grade {self.__grade}")

s1 = Student("maaz seraiki")
s2 = Student("zahran chisti")

s1.set_grade(85)
s2.set_grade(92)

s1.display_info()
s2.display_info()
```
