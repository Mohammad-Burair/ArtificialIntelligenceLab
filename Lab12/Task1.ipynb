```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("Mall_Customers.csv")

df.head()
```

```python
df = df.drop("CustomerID", axis=1)
```

```python
df["Gender"] = df["Gender"].map({"Male": 0, "Female": 1})
```

```python
X = df.values
```

```python
kmeans_no_scale = KMeans(n_clusters=5, random_state=42)

y_no_scale = kmeans_no_scale.fit_predict(X)
```

```python
print(kmeans_no_scale.cluster_centers_)
```

```python
X_scaled = df.copy()

age_column = X_scaled["Age"]
```

```python
scaler = StandardScaler()

X_scaled[["Gender", "Annual Income (k$)", "Spending Score (1-100)"]] = \
scaler.fit_transform(
    X_scaled[["Gender", "Annual Income (k$)", "Spending Score (1-100)"]]
)

X_scaled["Age"] = age_column
```

```python
X_scaled = X_scaled.values
```

```python
kmeans_scaled = KMeans(n_clusters=5, random_state=42)

y_scaled = kmeans_scaled.fit_predict(X_scaled)
```

```python
print(kmeans_scaled.cluster_centers_)
```
