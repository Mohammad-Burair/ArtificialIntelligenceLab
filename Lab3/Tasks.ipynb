```python
import random

class SecuritySystem:
    def __init__(self):
        self.components = {chr(65+i): random.choice(['safe', 'vulnerable']) for i in range(9)}

    def get_percept(self):
        return self.components

class SecurityAgent:
    def act(self, percept):
        for component, status in percept.items():
            if status == 'vulnerable':
                percept[component] = 'safe'
                print(f"{component} patched")

def run_agent(agent, env):
    print("initial state:", env.get_percept())
    agent.act(env.get_percept())
    print("final state:", env.get_percept())

if __name__ == "__main__":
    env = SecuritySystem()
    agent = SecurityAgent()
    run_agent(agent, env)
```

```python
import random

class datacenter:
    def __init__(self):
        self.servers = {f"server{i+1}": random.choice(['underloaded', 'balanced', 'overloaded']) for i in range(5)}

    def get_percept(self):
        return self.servers

    def display_state(self):
        print(self.servers)

class loadbalanceragent:
    def act(self, percept):
        print("initial server states:", percept)

        for server, status in percept.items():
            if status == 'overloaded':
                for target, t_status in percept.items():
                    if t_status == 'underloaded':
                        print(f"{server} -> {target}")
                        percept[server] = 'balanced'
                        percept[target] = 'balanced'
                        break

        print("final server states:", percept)

def run_agent(agent, env):
    percept = env.get_percept()
    agent.act(percept)

if __name__ == "__main__":
    env = datacenter()
    agent = loadbalanceragent()
    run_agent(agent, env)
```

```python
import random

class BackupSystem:
    def __init__(self):
        self.tasks = {f"task{i+1}": random.choice(['completed', 'failed']) for i in range(5)}

    def get_percept(self):
        return self.tasks

class BackupAgent:
    def act(self, percept):
        for task, status in percept.items():
            if status == 'failed':
                percept[task] = 'completed'
                print(f"{task} retried and completed")

def run_agent(agent, env):
    print("initial tasks:", env.get_percept())
    agent.act(env.get_percept())
    print("final tasks:", env.get_percept())

if __name__ == "__main__":
    env = BackupSystem()
    agent = BackupAgent()
    run_agent(agent, env)
```

```python
import random

class SecuritySystem:
    def __init__(self):
        self.components = {
            'A': random.choice(['safe', 'low', 'high']),
            'B': random.choice(['safe', 'low', 'high']),
            'C': random.choice(['safe', 'low', 'high']),
            'D': random.choice(['safe', 'low', 'high']),
            'E': random.choice(['safe', 'low', 'high']),
            'F': random.choice(['safe', 'low', 'high']),
            'G': random.choice(['safe', 'low', 'high']),
            'H': random.choice(['safe', 'low', 'high']),
            'I': random.choice(['safe', 'low', 'high'])
        }

    def get_percept(self):
        return self.components

class UtilitySecurityAgent:
    def act(self, percept):
        for comp, status in percept.items():
            if status == 'low':
                percept[comp] = 'safe'

def run_agent(agent, env):
    print("initial state:", env.get_percept())
    agent.act(env.get_percept())
    print("final state:", env.get_percept())

if __name__ == "__main__":
    env = SecuritySystem()
    agent = UtilitySecurityAgent()
    run_agent(agent, env)
```

```python
class Hospital:
    def __init__(self):
        self.rooms = {
            101: 'paracetamol',
            102: 'insulin',
            103: 'antibiotic'
        }
        self.staff_available = True

    def get_percept(self):
        return self.rooms, self.staff_available

class DeliveryRobot:
    def act(self, percept):
        rooms, staff = percept
        for room, medicine in rooms.items():
            print(f"moving to room {room}")
            print(f"scanning patient id in room {room}")
            print(f"delivering {medicine} to room {room}")
            if not staff:
                print(f"alert staff for room {room}")

def run_agent(agent, env):
    percept = env.get_percept()
    agent.act(percept)

if __name__ == "__main__":
    env = Hospital()
    robot = DeliveryRobot()
    run_agent(robot, env)
```

```python
class Building:
    def __init__(self):
        self.rooms = {
            'a': 'safe', 'b': 'safe', 'c': 'fire',
            'd': 'safe', 'e': 'fire', 'f': 'safe',
            'g': 'safe', 'h': 'safe', 'j': 'fire'
        }

class FireRobot:
    def act(self, env, path):
        for room in path:
            if env.rooms[room] == 'fire':
                env.rooms[room] = 'safe'
                print(f"fire in {room} extinguished")
            else:
                print(f"{room} is safe")

def run_agent(agent, env, path):
    agent.act(env, path)
    print("final environment:", env.rooms)

if __name__ == "__main__":
    env = Building()
    robot = FireRobot()
    path = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'j']
    run_agent(robot, env, path)
```
