```python
environment = {
    "Zone A": [3, 5],
    "Zone B": [2, 9]
}


def thief_min(options):
    print("  thief evaluating hiding options:", options)
    choice = min(options)
    print("  thief chooses minimum value =", choice)
    return choice


def robot_max(env):
    print("robot evaluating zones:\n")

    results = {}

    for zone in env:
        print("robot moves to:", zone)

        # thief reacts
        value = thief_min(env[zone])

        results[zone] = value
        print("state value for", zone, "=", value)

    best_zone = max(results, key=results.get)

    print("\nrobot chooses best zone:", best_zone)
    print("optimal safety value:", results[best_zone])

    return best_zone


robot_max(environment)
```

```python
import math

game_tree = {
    "attack": {
        "a1": [5, 6],
        "a2": [7, 4]
    },
    "defend": {
        "d1": [3, 8],
        "d2": [6, 2]
    },
    "gather": {
        "g1": [1, 9],
        "g2": [4, 7]
    }
}

def min_value(values, alpha, beta, node_name):
    print(f"\n  min node: {node_name}")

    v = math.inf

    for val in values:
        print("    checking:", val)

        v = min(v, val)
        beta = min(beta, v)

        print(f"    v={v}, alpha={alpha}, beta={beta}")

        if beta <= alpha:
            print("pruned remaining branches in", node_name)
            break

    return v


def max_value(tree):
    alpha = -math.inf
    beta = math.inf

    results = {}

    visited_leaves = 0

    print("starting minimax with alpha-beta pruning\n")

    for move in tree:
        print("max node:", move)

        min_results = {}

        for child in tree[move]:
            value = min_value(tree[move][child], alpha, beta, child)
            min_results[child] = value
            visited_leaves += 2  

        best_for_move = max(min_results.values())
        results[move] = best_for_move

        alpha = max(alpha, best_for_move)

        print(f"\nresult of {move} = {best_for_move}")
        print(f"updated alpha = {alpha}")

    best_move = max(results, key=results.get)

    print("final best move:", best_move)
    print("best value:", results[best_move])

    return visited_leaves

def minimax_only(tree):
    count = 0

    for move in tree:
        for child in tree[move]:
            count += len(tree[move][child])

    return count


print("\nalpha-beta ")
alpha_beta_nodes = max_value(game_tree)

print("\ncomparison ")
minimax_nodes = minimax_only(game_tree)

print("minimax nodes evaluated:", minimax_nodes)
print("alpha-beta nodes visited:", alpha_beta_nodes)

print("nodes saved:", minimax_nodes - alpha_beta_nodes)
```
